Q2. Customer Spending Analysis Using the Tips Dataset
Use Seaborn's built-in tips dataset.

(a) Calculate the mean, median, and all modes of total_bill and tip. Then divide the dataset into Lunch and Dinner groups and calculate the mean and median bill for each group.

(b) Identify the three largest bills, temporarily exclude them, and recalculate the overall mean and median. Use Python to compare the results before and after exclusion, and explain which measure better represents typical customer spending.

```python
import seaborn as sns
import pandas as pd

df = sns.load_dataset("tips")

print(df.head())
```

```python
def final_mean(data):
    return sum(data) / len(data)


def final_median(data):
    data = sorted(data)
    n = len(data)

    if n % 2 == 1:
        return data[n // 2]
    else:
        mid = n // 2
        return (data[mid - 1] + data[mid]) / 2


def final_modes(data):
    frequency = {}

    for value in data:
        frequency[value] = frequency.get(value, 0) + 1

    max_frequency = max(frequency.values())

    modes = []
    for value in frequency:
        if frequency[value] == max_frequency:
            modes.append(value)

    return modes
```

```python
for column in ["total_bill", "tip"]:

    data = df[column].tolist()

    print("\nColumn:", column)
    print("Mean:", final_mean(data))
    print("Median:", final_median(data))
    print("Mode:", final_mode(data))
```

```python
# Lunch group
lunch = df[df["time"] == "Lunch"]["total_bill"].tolist()

# Dinner group
dinner = df[df["time"] == "Dinner"]["total_bill"].tolist()

print("Lunch Mean:", final_mean(lunch))
print("Lunch Median:", final_median(lunch))

print("Dinner Mean:", final_mean(dinner))
print("Dinner Median:", final_median(dinner))
```

Q4. Investigating the Effect of Mathematical Transformations
A researcher records these measurements:

measurements = [12, 15, 18, 20, 25, 30, 40]
(a) Create two new datasets: one by adding 10 to every observation and another by multiplying every observation by 4. Calculate the mean, population variance, and population standard deviation for all three datasets. Use Python to verify the mathematical relationships between the original and transformed statistics.

(b) Create a fourth dataset by multiplying every observation by \(-2\). Calculate its mean, variance, and standard deviation. Explain why a negative multiplier changes the direction of the values but does not make variance or standard deviation negative.

```python

measurements = [12, 15, 18, 20, 25, 30, 40]


data_add_10 = [x + 10 for x in measurements]
data_multiply_4 = [x * 4 for x in measurements]
data_multiply_minus_2 = [x * (-2) for x in measurements]

def calculate_mean(data):
    total = 0

    for x in data:
        total += x

    return total / len(data)


def calculate_variance(data):
    mean = calculate_mean(data)
    total = 0

    for x in data:
        total += (x - mean) ** 2

    return total / len(data)


def calculate_std(data):
    variance = calculate_variance(data)
    return variance ** 0.5

datasets = {
    "Original": measurements,
    "Add 10": data_add_10,
    "Multiply by 4": data_multiply_4,
    "Multiply by -2": data_multiply_minus_2
}


print("=" * 65)
print("       MATHEMATICAL TRANSFORMATION REPORT")
print("=" * 65)

for name, data in datasets.items():
    print("\nDataset:", name)
    print("Values:", data)
    print("Mean:", round(calculate_mean(data), 4))
    print("Population Variance:", round(calculate_variance(data), 4))
    print("Population Standard Deviation:", round(calculate_std(data), 4))

print("=" * 65)
```

Q5. Outlier Detection and Robust Spread Analysis
A data analyst records daily sales, in thousands of rupees.

sales = [    12, 14, 15, 15, 16, 17, 18,    19, 20, 21, 22, 24, 25, 120]
(a) Write Python code to calculate the mean, median, population variance, and population standard deviation. Then identify potential outliers using the IQR method and report the quartiles, IQR, lower bound, upper bound, and flagged observations.

(b) Temporarily exclude the flagged outlier and recalculate the mean, median, variance, and standard deviation. Compare the percentage changes in variance and standard deviation, and recommend which statistics best describe typical daily sales while explaining why an extreme observation should be investigated before removal.

```python
sales = [12, 14, 15, 15, 16, 17, 18,
         19, 20, 21, 22, 24, 25, 120]


# Mean function
def calculate_mean(data):
    total = 0
    for x in data:
        total += x
    return total / len(data)
```

```python
calculate_mean(data)
```

```python
def calculate_median(data):
    arr = sorted(data)
    n = len(arr)
    mid = n // 2

    if n % 2 == 1:
        return arr[mid]
    else:
        return (arr[mid - 1] + arr[mid]) / 2
```

```python
calculate_median(data)
```

```python
def calculate_variance(data):
    mean = calculate_mean(data)
    total = 0

    for x in data:
        total += (x - mean) ** 2

    return total / len(data)
```

```python
calculate_variance(data)
```

```python
def calculate_std(data):
    return calculate_variance(data) ** 0.5
```

```python
calculate_std(data)
```

```python
def calculate_quartiles(data):
    arr = sorted(data)
    n = len(arr)
    mid = n // 2

    lower_half = arr[:mid]
    upper_half = arr[mid:] if n % 2 == 0 else arr[mid + 1:]

    q1 = calculate_median(lower_half)
    q3 = calculate_median(upper_half)

    return q1, q3
```

```python
calculate_quartiles(data)
```

```python
q1, q3 = calculate_quartiles(sales)
iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

outliers = []

for x in sales:
    if x < lower_bound or x > upper_bound:
        outliers.append(x)
```
